```python
text_cols = ["title", "track_artist", "genre_l1", "genre_l2"]

for c in text_cols:
    s = df[c].dropna()
    print(c,
          "| extra spaces:", (s != s.str.strip()).sum(),
          "| unique values:", s.nunique(),
          "| unique after fixing case/spaces:", s.str.strip().str.lower().nunique())
```

```python
print(df["genre_l1"].value_counts())
```

```python
before_unique = df["track_artist"].nunique()

df["track_artist"] = df["track_artist"].str.strip()

after_unique = df["track_artist"].nunique()
print("Unique artists before:", before_unique, "| after:", after_unique)

changed = raw["track_artist"].notna() & (raw["track_artist"] != raw["track_artist"].str.strip())
print("Rows changed:", changed.sum())

pd.DataFrame({
    "before": raw.loc[changed, "track_artist"].map(repr),
    "after": df.loc[changed, "track_artist"].map(repr),
}).head(10)
```

```python
print("Exact duplicate rows:", df.duplicated().sum())
print("Repeated track_ids:", df["track_id"].duplicated().sum())

same_song = df["title_normalized"].notna() & df.duplicated(subset=["title_normalized", "track_artist"], keep=False)
print("Rows that share a title + artist with another row:", same_song.sum())
```

```python
cols_to_see = ["track_id", "title", "track_artist", "release_year", "genre_l1", "track_popularity", "duration_ms"]
df[same_song].sort_values(["title_normalized", "track_artist"])[cols_to_see].head(20)
```

```python
key = pd.DataFrame({
    "title": df["title"].str.strip().str.lower(),
    "artist": df["track_artist"],
})
has_both = key["title"].notna() & key["artist"].notna()

possible_dup = has_both & key.duplicated(keep=False)
extra_copies = has_both & key.duplicated()

print("Rows sharing exact title + artist:", possible_dup.sum())
print("Extra copies beyond the first:", extra_copies.sum())
```

```python
df["possible_duplicate"] = possible_dup
print("Rows:", len(df))
print(df["possible_duplicate"].value_counts())
```

```python
missing_before = raw.isna().sum()
print("Missing values in the original data:")
print(missing_before[missing_before > 0])

print()
print("Hidden missing values:")
print("genre_l1 = 'Unknown':", (df["genre_l1"] == "Unknown").sum())
print("genre_l2 = 'Unknown':", (df["genre_l2"] == "Unknown").sum())
print("Empty titles:", (df["title"].str.strip() == "").sum())
print("Empty artists:", (df["track_artist"] == "").sum())
```

```python
print(df["streams_source"].value_counts())
print()
print(pd.crosstab(df["streams_total_estimated"], df["streams_source"].isna(),
                  rownames=["estimated?"], colnames=["source missing?"]))
```

```python
for c in ["genre_l1", "genre_l2"]:
    df[c] = df[c].mask(df[c] == "Unknown")

print("Same rows missing both genres:", (df["genre_l1"].isna() == df["genre_l2"].isna()).all())

comparison = pd.DataFrame({
    "missing_before": raw.isna().sum(),
    "missing_after": df[raw.columns].isna().sum(),
})
comparison[(comparison["missing_before"] > 0) | (comparison["missing_after"] > 0)]
```

```python
num_cols = ["release_year", "track_popularity", "streams_total", "duration_ms",
            "danceability", "energy", "valence", "acousticness",
            "instrumentalness", "loudness", "tempo"]

df[num_cols].describe().T
```

```python
checks = {
    "release_year before 1900 or after 2026": ~df["release_year"].between(1900, 2026),
    "popularity outside 0-100": ~df["track_popularity"].between(0, 100),
    "negative streams": df["streams_total"] < 0,
    "duration under 30 seconds": df["duration_ms"] < 30_000,
    "duration over 20 minutes": df["duration_ms"] > 20 * 60 * 1000,
    "tempo = 0": df["tempo"] == 0,
    "tempo over 250 BPM": df["tempo"] > 250,
    "loudness above 0 dB": df["loudness"] > 0,
}
for c in ["danceability", "energy", "valence", "acousticness", "instrumentalness"]:
    checks[f"{c} outside 0-1"] = ~df[c].between(0, 1)

for name, mask in checks.items():
    print(f"{name}: {mask.sum()}")
```

```python
short = df["duration_ms"] < 30_000
long = df["duration_ms"] > 20 * 60 * 1000
show = ["title", "track_artist", "genre_l1", "duration_ms", "track_popularity"]

print("SHORT TRACKS (examples):")
print(df.loc[short, show].sample(8, random_state=1))
print()
print("LONG TRACKS (examples):")
print(df.loc[long, show].sample(8, random_state=1))
print()
print("Max values:")
print(df[num_cols].max())
```

```python
df["unusual_duration"] = short | long
print("Flagged unusual duration:", df["unusual_duration"].sum())
print("Rows:", len(df))
```

```python
summary = pd.DataFrame({
    "raw": [
        len(raw), raw.shape[1],
        raw["track_popularity"].mean(), raw["streams_total"].sum(),
        raw["genre_l1"].nunique(), raw["genre_l1"].notna().sum(),
        raw["track_artist"].nunique(),
    ],
    "cleaned": [
        len(df), df.shape[1],
        df["track_popularity"].mean(), df["streams_total"].sum(),
        df["genre_l1"].nunique(), df["genre_l1"].notna().sum(),
        df["track_artist"].nunique(),
    ],
}, index=[
    "rows", "columns",
    "avg track_popularity", "total streams",
    "number of genres", "tracks with a genre",
    "unique artists",
])
summary["difference"] = summary["cleaned"] - summary["raw"]
pd.set_option("display.float_format", "{:,.2f}".format)
summary
```

```python
by_genre = df.groupby("genre_l1").agg(
    tracks=("track_id", "count"),
    avg_popularity=("track_popularity", "mean"),
)
print("Tracks counted in genre groups:", by_genre["tracks"].sum())
print("Tracks with no genre:", df["genre_l1"].isna().sum())
print("Total:", by_genre["tracks"].sum() + df["genre_l1"].isna().sum())
by_genre.sort_values("tracks", ascending=False)
```
